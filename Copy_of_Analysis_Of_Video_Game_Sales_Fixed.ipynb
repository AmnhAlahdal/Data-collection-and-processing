```
numeric_cols = df.select_dtypes(include=np.number).columns

print("Numeric Columns:")
print(numeric_cols)

print("\nMissing values in numeric columns:")
print(df[numeric_cols].isnull().sum())
```

```
# Mean Imputation (on a copy)
df_mean = df.copy()
for col in numeric_cols:
    if col == 'Year':
        df_mean[col] = df_mean[col].astype(float).fillna(df_mean[col].mean()).round().astype('Int64')
    else:
        df_mean[col] = df_mean[col].fillna(df_mean[col].mean())

print("Missing after Mean Imputation:")
print(df_mean[numeric_cols].isnull().sum())
```

```
# Median Imputation (on a copy)
df_median = df.copy()
for col in numeric_cols:
    if col == 'Year':
        df_median[col] = df_median[col].astype(float).fillna(df_median[col].median()).round().astype('Int64')
    else:
        df_median[col] = df_median[col].fillna(df_median[col].median())

print("Missing after Median Imputation:")
print(df_median[numeric_cols].isnull().sum())
```

```
# Constant Imputation (on a copy)
df_constant = df.copy()
for col in numeric_cols:
    df_constant[col] = df_constant[col].fillna(0)

print("Missing after Constant Imputation:")
print(df_constant[numeric_cols].isnull().sum())
```

```
# Categorical columns — fill missing values on main df
cat_cols = ["Platform", "Genre", "Publisher"]

# Mode fill
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

# Forward Fill + Backward Fill
df[cat_cols] = df[cat_cols].ffill()
df[cat_cols] = df[cat_cols].bfill()

print("Missing after categorical fill:")
print(df[cat_cols].isnull().sum())
```

```
# Drop remaining rows with any missing values (Year column)
df_cleaned = df.dropna()

print("Before dropping missing values:", len(df))
print("After dropping missing values:", len(df_cleaned))
```

# **🔹 Lab 4: Outlier Detection and Handling**

1️⃣ Detect outliers visually using Boxplot

```
# Use the main df (already cleaned and imputed)
sns.boxplot(x=df['Global_Sales'])
plt.title('Boxplot of Global Sales')
plt.show()
```

2️⃣ Detect outliers using IQR rule (1.5 IQR)

🟢 Two Variables: (Global_Sales) and (NA_Sales)

**(Global_Sales)**

```
# Calculate Q1 and Q3 for Global_Sales
Q1 = df['Global_Sales'].quantile(0.25)
Q3 = df['Global_Sales'].quantile(0.75)

print("Q1:", Q1)
print("Q3:", Q3)
```

```
# Calculate IQR
IQR = Q3 - Q1
print("IQR:", IQR)
```

```
# Calculate lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```
# Detect outliers
outliers = df[(df['Global_Sales'] < lower_bound) | (df['Global_Sales'] > upper_bound)]
print(outliers)
```

**(NA_Sales)**

```
# Calculate Q1 and Q3 for NA_Sales
Q1_na = df['NA_Sales'].quantile(0.25)
Q3_na = df['NA_Sales'].quantile(0.75)

IQR_na = Q3_na - Q1_na

lower_na = Q1_na - 1.5 * IQR_na
upper_na = Q3_na + 1.5 * IQR_na

print("Lower Bound:", lower_na)
print("Upper Bound:", upper_na)
```

```
# Detect outliers in NA_Sales
outliers_na = df[(df['NA_Sales'] < lower_na) | (df['NA_Sales'] > upper_na)]
print(outliers_na)
```

```
# Analyze outliers separately for Global_Sales
global_outliers = df[(df['Global_Sales'] < lower_bound) | (df['Global_Sales'] > upper_bound)]

print("Number of Global Sales Outliers:")
print(global_outliers.shape[0])
print(global_outliers[['Name','Platform','Year','Global_Sales']])
```

```
# Analyze outliers separately for NA_Sales
na_outliers = df[(df['NA_Sales'] < lower_na) | (df['NA_Sales'] > upper_na)]

print("Number of NA Sales Outliers:")
print(na_outliers.shape[0])
print(na_outliers[['Name','Platform','Year','NA_Sales']])
```

```
# Outlier Examination
print("Outlier Examination:")
print("The detected outliers were examined to understand their cause.")
print("Some values may be due to data entry errors.")
print("Some may represent real extreme values.")
print("Others may be rare but interesting observations in the dataset.")
```

```
# Handle outliers for all sales columns — apply to main df
sales_cols = ['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales', 'Global_Sales']

for col in sales_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    print(f"\nColumn: {col}")
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
```

```
for col in sales_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lb = Q1 - 1.5 * IQR
    ub = Q3 + 1.5 * IQR
    outliers_col = df[(df[col] < lb) | (df[col] > ub)]
    print(f"\nColumn: {col}")
    print("Lower Bound:", lb)
    print("Upper Bound:", ub)
    print("Number of Outliers:", len(outliers_col))
```

```
for col in sales_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lb = Q1 - 1.5 * IQR
    ub = Q3 + 1.5 * IQR
    outliers_col = df[(df[col] < lb) | (df[col] > ub)]
    print(f"\nSample outliers in {col}:")
    print(outliers_col[[col, 'Name']].head())
```

```
# ── Method 1: Removal ──
df_clean = df.copy()
for col in sales_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1
    lb = Q1 - 1.5 * IQR
    ub = Q3 + 1.5 * IQR
    df_clean = df_clean[(df_clean[col] >= lb) & (df_clean[col] <= ub)]

print("Dataset size before removing outliers:", df.shape)
print("Dataset size after removing all sales outliers:", df_clean.shape)
```

```
# ── Method 2: Capping ──
Q1_g = df['Global_Sales'].quantile(0.25)
Q3_g = df['Global_Sales'].quantile(0.75)
IQR_g = Q3_g - Q1_g
lb_g = Q1_g - 1.5 * IQR_g
ub_g = Q3_g + 1.5 * IQR_g

df_capped = df.copy()
df_capped['Global_Sales'] = df_capped['Global_Sales'].clip(lower=lb_g, upper=ub_g)

print("After Capping - Max Global_Sales:", df_capped['Global_Sales'].max())
print("After Capping - Min Global_Sales:", df_capped['Global_Sales'].min())

# ── Method 3: Log Transformation ──
df_log = df.copy()
df_log['Global_Sales_Log'] = np.log1p(df_log['Global_Sales'])

print("\nAfter Log Transformation:")
print(df_log['Global_Sales_Log'].describe())

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(df['Global_Sales'], bins=50, color='steelblue')
axes[0].set_title("Before Log Transformation")
axes[0].set_xlabel("Global Sales")

axes[1].hist(df_log['Global_Sales_Log'], bins=50, color='coral')
axes[1].set_title("After Log Transformation")
axes[1].set_xlabel("Log of Global Sales")

plt.tight_layout()
plt.show()

print("\nWhy values are considered outliers:")
print("Values beyond 1.5*IQR from Q1 and Q3 are statistically considered outliers.")
print("\nWhy we applied multiple methods:")
print("- Removal: Simple and effective but loses data.")
print("- Capping: Keeps all data but limits extreme values.")
print("- Log Transformation: Reduces skewness while preserving all data points.")
```

```
# ✅ Apply capping to the main df so downstream labs use clean data
for col in sales_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lb = Q1 - 1.5 * IQR
    ub = Q3 + 1.5 * IQR
    df[col] = df[col].clip(lower=lb, upper=ub)

print("✅ Outliers capped on main df")
print(df.shape)
```

### Wrapper Method (RFE)

```
from sklearn.feature_selection import RFE
from sklearn.linear_model import LinearRegression

# Work on a copy — do NOT overwrite df
df_rfe = df.copy()

categorical_cols_rfe = ['Name', 'Platform', 'Genre', 'Publisher']

# Fill missing values
df_rfe[categorical_cols_rfe] = df_rfe[categorical_cols_rfe].fillna('Unknown')
df_rfe = df_rfe.fillna(0)

# Encode categorical
le_rfe = LabelEncoder()
df_encoded = df_rfe.copy()
for col in categorical_cols_rfe:
    df_encoded[col] = le_rfe.fit_transform(df_encoded[col].astype(str))

# Define target and features
y = df_encoded['Global_Sales']
X = df_encoded.drop('Global_Sales', axis=1)

# Build RFE with LinearRegression
model = LinearRegression()
rfe = RFE(estimator=model, n_features_to_select=3)
rfe.fit(X, y)

print("Selected Features (Support):")
print(rfe.support_)

print("\nFeature Ranking:")
print(rfe.ranking_)

selected_features = X.columns[rfe.support_]
print("\nBest 3 Features:")
print(selected_features)
```

Remove punctuation and emojis

Stopwords removing

## 3. Stack() & Unstack()

```
# Stack/Unstack example with game data
data_games = {
    'Name': ['Mario Kart', 'FIFA Series', 'Call of Duty', 'The Legend of Zelda'],
    'Platforms': [['Wii', 'DS', 'Switch'], ['PS4', 'Xbox One', 'PC'], ['PS4', 'Xbox One'], ['NES', 'SNES', 'N64', 'Switch']],
    'Global_Sales': [35.82, 28.50, 30.00, 19.01]
}

df_games = pd.DataFrame(data_games)

# Unstack
df_unstacked = df_games.set_index(['Name', 'Global_Sales']).unstack(fill_value=None)
print("Unstacked Result:")
print(df_unstacked)
```

```
# Stack
stacked_df = df_unstacked.stack(future_stack=True).reset_index()
stacked_df.columns = ['Platform', 'Genre', 'Value']

print("Formatted Stack Result:")
print(stacked_df)
```

### Explode() - Splits list values into separate rows

```
data_explode = {
    'Name': ['Mario Kart', 'FIFA Series', 'Call of Duty', 'The Legend of Zelda'],
    'Platforms': [
        ['Wii', 'DS', 'Switch'],
        ['PS4', 'Xbox One', 'PC'],
        ['PS4', 'Xbox One'],
        ['NES', 'SNES', 'N64', 'Switch']
    ],
    'Global_Sales': [35.82, 28.50, 30.00, 19.01]
}

df_lists = pd.DataFrame(data_explode)

print("Before explode():")
print(df_lists)

# Explode Platforms into separate rows
df_exploded = df_lists.explode('Platforms').reset_index(drop=True)

print("\nAfter explode():")
print(df_exploded)
```
